# typecasting 
by (astype() method)
we can explicitly convert or cast an array from one dtype to another using ndarray’s astype method:

Calling astype always creates a new array (a copy of the data), even if
the new dtype is the same as the old dtype.

```python
arr = np.array([1,2,3,4])
arr.dtype
```

```python
float_arr=arr.astype(np.float64)
float_arr.dtype
```

```python
float_arr1= arr3.astype(np.int32)
float_arr1.dtype
```

```python
arr = np.array([1.4,-2.9,0.3,8.4])
arr
arr.astype(np.int32)
```

```python
numeric_strings = np.array(['1.25', '-9.6', '4.2'], dtype=np.bytes_)
numeric_strings=numeric_strings.astype(float)
numeric_strings
```

```python
arr5= np.array([1.4,-2.9,0.3,8.4])
arr5
arr5.astype(np.bytes_)
```

```python
int_array = np.arange(10)
calibers = np.array([.22, .270, .357, .380, .44, .50], dtype=np.float64)
int_array.astype(calibers.dtype)
```

```python
empty_uint32 = np.empty(8, dtype='u4')
empty_uint32
```
